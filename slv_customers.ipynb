```python
import pyspark.sql.functions as F
import pyspark.sql.types as T
```

```python
df = spark.read.table('olist_dataset.bronze.customers')
display(df.limit(10))
```

```python
df = df.withColumn('city', F.initcap(F.col('city')))
```

```python
@F.udf(returnType=T.StringType())
def short_to_long(x : str) :
    match x :
        case "AC": return "Acre"
        case "AL": return "Alagoas"
        case "AM": return "Amazonas"
        case "AP": return "Amapá"
        case "BA": return "Bahia"
        case "CE": return "Ceará"
        case "DF": return "Distrito Federal"
        case "ES": return "Espírito Santo"
        case "GO": return "Goiás"
        case "MA": return "Maranhão"
        case "MG": return "Minas Gerais"
        case "MS": return "Mato Grosso do Sul"
        case "MT": return "Mato Grosso"
        case "PA": return "Pará"
        case "PB": return "Paraíba"
        case "PE": return "Pernambuco"
        case "PI": return "Piauí"
        case "PR": return "Paraná"
        case "RJ": return "Rio de Janeiro"
        case "RN": return "Rio Grande do Norte"
        case "RO": return "Rondônia"
        case "RR": return "Roraima"
        case "RS": return "Rio Grande do Sul"
        case "SC": return "Santa Catarina"
        case "SE": return "Sergipe"
        case "SP": return "São Paulo"
        case "TO": return "Tocantins"
        case _: return "Unknown"
```

```python
df = df.withColumn('state',short_to_long(F.col("state")).cast('string'))

display(df.limit(5))
```

```python
display(df)
```

```python
df = df.drop('ingested_at','source_file')
```

```python
df.write.format('delta')\
    .mode('overwrite')\
    .option('mergeSchema',True)\
    .saveAsTable('olist_dataset.silver.customers')
```
