```python
from pyspark.sql.functions import col
import pyspark.sql.functions as F
import pyspark.sql.types as T
```

```python
df = spark.read.table('olist_dataset.bronze.products')

display(df.limit(10))
```

```python
cat_names = spark.read.table('olist_dataset.bronze.category')

display(cat_names.limit(10))
```

```python
df_with_eng_cat = df.join(cat_names,on=df['product_category_name'] == cat_names['category_in_brazilian'],how= "left")
```

```python
final_df = df_with_eng_cat.select('product_id','category_in_english','product_name_lenght','product_description_lenght','product_photos_qty','product_weight_g','product_length_cm','product_height_cm','product_width_cm')

display(final_df.limit(5))
```

```python
renamed_final_df = final_df.withColumnRenamed('category_in_english','category')
```

```python
display(renamed_final_df)
```

```python
renamed_final_df.write.format('delta')\
    .option('mergeSchema',True)\
    .mode('overwrite')\
    .saveAsTable('olist_dataset.silver.products')
```
