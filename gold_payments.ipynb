```python
%sql
create or replace temporary view pay as

select 
    p.order_id,
    p.payment_sequential,
    p.payment_type,
    p.payment_installments,
    p.payment_value,
    cast(o.purchase_time as date) as payment_date,
    date_format(o.purchase_time, 'HH:mm:ss') as payment_time,
    quarter,
    year
from olist_dataset.silver.payments as p
left join olist_dataset.silver.delivered_orders as o
on p.order_id = o.order_id
```
