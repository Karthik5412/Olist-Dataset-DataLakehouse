```python
%sql
CREATE OR REPLACE TEMPORARY VIEW cost AS
SELECT 
    *,
    (price + shipping_cost) AS  total_price
FROM olist_dataset.silver.items
```

```python
%sql
CREATE OR REPLACE TEMPORARY VIEW reviews AS
SELECT 
    item_id,
    round(avg(review_score),2) AS overall_review ,
    count(has_reviewed) filter(WHERE has_reviewed == true ) AS total_reviews
FROM olist_dataset.silver.items
GROUP BY item_id
```

```python
%sql
SELECT 
    c.order_id,
    c.product_id,
    c.seller_id,
    c.item_id,
    c.shipping_date as est_shipping_date,
    c.price,
    c.shipping_cost,
    c.total_price,
    c.review_score,
    c.has_reviewed,
    r.overall_review,
    r.total_reviews
FROM cost AS c 
LEFT JOIN reviews AS r 
ON c.item_id = r.item_id
```
